```python
import pandas as pd
```

1.Créer une series contenant 4 temperatures

```python
temperatures = pd.Series([15, 28, 32, 44, 35])
print(temperatures)
```

2.Afficher son index 

```python
print(temperatures.index)
```

3.Afficher la 1ère valeur

```python
print(temperatures.iloc[0])
```

4.Afficher la dernière valeur 

```python
print(temperatures.iloc[-1])
```

Créer, de deux façon différentes, une Series de 4 températures avec comme index les heures de 
mesure 12h, 13h, 14h et 15h 

```python
temperatures_heures = pd.Series(
    [24.5, 25.3, 26.1, 27.4],
    index=["12h", "13h", "14h", "15h"]
)

print(temperatures_heures)
```

```python
temperatures_heures = pd.Series([24.5, 25.3, 26.1, 27.4])

temperatures_heures.index = ["12h", "13h", "14h", "15h"]

print(temperatures_heures)
```

6.Afficher la 1ère et la dernière valeur 

```python
print("Première valeur :", temperatures_heures.iloc[0])
print("Dernière valeur :", temperatures_heures.iloc[-1])
```
